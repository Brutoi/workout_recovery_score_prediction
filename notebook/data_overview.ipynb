```python
# Get the count of each unique values of each features
# for checking garbage values

for i in df.select_dtypes(include=["str", "object"]).columns:
    print(df[i].value_counts())
    print("***"*10)
```

```python
# Get all the unique values of each features
# for checking inconsistent values

columns = df.select_dtypes(include=["object", "str"])

for col in columns:
    print(f"============ feature {col} ============")
    print(df[col].unique())
```

```python
# Unique Values of each features

print("********************** CATEGORICAL FEATURES **********************")
for col in df.select_dtypes(include=["object", "str"]).columns:
    print(df[col].value_counts())
    print(f"total unique values: {len(df[col].dropna().unique())}")
    print("=====================================\n")

print("********************** NUMERICAL FEATURES **********************")
for col in df.select_dtypes(include=["number"]).columns:
    print(df[col].value_counts())
    print(f"total unique values: {len(df[col].dropna().unique())}")
    print(f"max value: {df[col].max()}")
    print(f"min value: {df[col].min()}")
    print("=====================================\n")
```
